# Article Df

```python
import pickle 
with open('../data/batch_collection_annenberg_html.pkl', 'rb') as f:
    all_article_pages = pickle.load(f)
```

```python
import pandas as pd 
article_df = pd.read_csv('../data/batch_article_text.csv')
```

```python
import os 
os.environ['OPENAI_API_KEY'] = open('/Users/spangher/.openai-bloomberg-project-key.txt').read().strip()
from openai import OpenAI
client = OpenAI()

def prompt_openai(prompt, model='gpt-4o-mini'):
    completion = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": prompt}]
    )
    return completion.choices[0].message.content
```

```python
CLEAN_ARTICLE_TEXT_PROMPT = """
You are a helpful editor assistant. I scraped this news article from the web and removed HTML tags, however, 
there is a lot of extraneous text from the HTML page that remains. Please only extract text related to the news article.
Do NOT miss any article text. 

Please preserve the paragraph breaks. 

Here is the news article:

<article>
{article_text}
</article>

Return only the CLEANED text of the news article, nothing else.

Your response:
"""

CLEAN_ARTICLE_TEXT_FROM_HTML_PROMPT = """
You are a helpful editor assistant. Here is an HTML page I scraped from the web.
Please only extract text related to the news article. Do NOT miss any article text. 

Please preserve all of the paragraph breaks. 

Here is the news article html:

<article html>
{article_html}
</article html>

Return only the CLEANED text of the news article, with \n\n in between each paragraph break, nothing else.

Your response:
"""
```

```python
prompt = CLEAN_ARTICLE_TEXT_PROMPT.format(article_text=article_df['article_text'].iloc[0])
prompt = CLEAN_ARTICLE_TEXT_FROM_HTML_PROMPT.format(article_html=all_article_pages[0])
```

```python
cleaned_article = prompt_openai(prompt)
```

```python
paragraphs = cleaned_article.split('\n\n')
```

# Batch Jobs: Parsing with BeautifulSoup + Cleaning

```python
import tiktoken
from tqdm.auto import tqdm
from bs4 import BeautifulSoup
def remove_tags(html):
    soup = BeautifulSoup(html, "lxml")
    tags_to_remove = ['script', 'style', 'noscript', 'meta', 'link', 'iframe', 'footer', 'header', 'nav', 'form', 'hr', 'button', 'svg', 'img', 'picture', 'source']
    
    # Remove them
    for tag in soup(tags_to_remove):
        tag.decompose()
    return str(soup)
```

```python
enc = tiktoken.get_encoding("o200k_base")
```

```python
len(enc.encode(all_article_pages[0]))
```

```python
all_prompts = []
tokens = []
for html in tqdm(all_article_pages):
    stripped_html = remove_tags(html)
    prompt = CLEAN_ARTICLE_TEXT_FROM_HTML_PROMPT.format(article_html=stripped_html) 
    all_prompts.append(prompt)
    tokens.append(enc.encode(prompt))
```

```python
all_tokens = sum(list(map(len, tokens)))
all_tokens / 1_000_000 * 0.150 
```

```python
output = prompt_openai(prompt)
```

```python
len(enc.encode(output))
```

```python
all_batched_prompts = []
for prompt in all_prompts:
    all_batched_prompts.append({
        "custom_id": "request-1",
        "method": "POST",
        "url": "/v1/chat/completions",
        "body": {
            "model": "gpt-4o-mini", 
            "messages": [{"role": "user", "content": "Hello world!"}]
            "max_tokens": 1000
        }
    })
```
